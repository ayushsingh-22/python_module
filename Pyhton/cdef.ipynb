1. Write code to implement a min-min program.


```python
def min_min_scheduling(exec_time):
    num_tasks = len(exec_time)
    num_resources = len(exec_time[0])
    
    resource_completion_time = [0] * num_resources
    task_allocation = [-1] * num_tasks

    for _ in range(num_tasks):
        min_time = float('inf')
        selected_task = -1
        selected_resource = -1

        for task in range(num_tasks):
            if task_allocation[task] == -1:  
                for resource in range(num_resources):

                    completion_time = resource_completion_time[resource] + exec_time[task][resource]
                    if completion_time < min_time:
                        min_time = completion_time
                        selected_task = task
                        selected_resource = resource

        task_allocation[selected_task] = selected_resource
        resource_completion_time[selected_resource] += exec_time[selected_task][selected_resource]

    return task_allocation, resource_completion_time

execution_times = [
    [5, 11, 15],  
    [8, 14, 13],   
    [12, 10, 20],  
    [14, 2, 22]   
]
task_allocation, resource_completion_time = min_min_scheduling(execution_times)

print("Task Allocation:", task_allocation)
print("Resource Completion Time:", resource_completion_time)
```

5. Write code to implement an Ant Colony Optimization (ACO) algorithm .

```python
import numpy as np
import random

class AntColonyOptimization:
    def __init__(self, num_ants, num_nodes, alpha, beta, rho, q, iterations, distances):
        self.num_ants = num_ants
        self.num_nodes = num_nodes
        self.alpha = alpha
        self.beta = beta
        self.rho = rho
        self.q = q
        self.iterations = iterations
        self.distances = distances
        self.pheromones = np.ones((num_nodes, num_nodes)) / num_nodes
        self.best_distance = float('inf')
        self.best_path = []

    def run(self):
        for _ in range(self.iterations):
            all_paths = self.construct_paths()
            self.update_pheromones(all_paths)
        return self.best_path, self.best_distance

    def construct_paths(self):
        all_paths = []
        for _ in range(self.num_ants):
            path = []
            visited = set()
            current_node = random.randint(0, self.num_nodes - 1)
            path.append(current_node)
            visited.add(current_node)
            for _ in range(self.num_nodes - 1):
                next_node = self.select_next_node(current_node, visited)
                path.append(next_node)
                visited.add(next_node)
                current_node = next_node
            path.append(path[0])
            all_paths.append((path, self.calculate_distance(path)))
            if self.calculate_distance(path) < self.best_distance:
                self.best_distance = self.calculate_distance(path)
                self.best_path = path
        return all_paths

    def select_next_node(self, current_node, visited):
        probabilities = []
        for next_node in range(self.num_nodes):
            if next_node not in visited:
                pheromone = self.pheromones[current_node][next_node] ** self.alpha
                distance = (1 / self.distances[current_node][next_node]) ** self.beta
                probabilities.append(pheromone * distance)
            else:
                probabilities.append(0)
        probabilities = probabilities / np.sum(probabilities)
        return np.random.choice(range(self.num_nodes), p=probabilities)

    def calculate_distance(self, path):
        total_distance = 0
        for i in range(len(path) - 1):
            total_distance += self.distances[path[i]][path[i + 1]]
        return total_distance

    def update_pheromones(self, all_paths):
        self.pheromones *= (1 - self.rho)
        for path, distance in all_paths:
            for i in range(len(path) - 1):
                from_node, to_node = path[i], path[i + 1]
                self.pheromones[from_node][to_node] += self.q / distance
                self.pheromones[to_node][from_node] += self.q / distance

distances = [
    [0, 2, 2, 12, 7],
    [2, 0, 4, 8, 2],
    [2, 4, 0, 1, 4],
    [5, 8, 1, 0, 2],
    [22, 12, 3, 12, 0]
]

aco = AntColonyOptimization(num_ants=5, num_nodes=5, alpha=1, beta=2, rho=0.5, q=100, iterations=100, distances=distances)
best_path, best_distance = aco.run()

print("Best Path:", best_path)
print("Best Distance:", best_distance)
```

6. Write code to implement a round robin algorithm .

```python
def round_robin(processes, burst_time, quantum):
    n = len(processes)
    remaining_burst = burst_time[:]
    waiting_time = [0] * n
    turnaround_time = [0] * n
    time = 0

    while True:
        done = True
        for i in range(n):
            if remaining_burst[i] > 0:
                done = False
                if remaining_burst[i] > quantum:
                    time += quantum
                    remaining_burst[i] -= quantum
                else:
                    time += remaining_burst[i]
                    waiting_time[i] = time - burst_time[i]
                    remaining_burst[i] = 0
        if done:
            break

    for i in range(n):
        turnaround_time[i] = burst_time[i] + waiting_time[i]

    avg_waiting_time = sum(waiting_time) / n
    avg_turnaround_time = sum(turnaround_time) / n

    return waiting_time, turnaround_time, avg_waiting_time, avg_turnaround_time

processes = [1, 2, 3, 4]
burst_time = [21, 3, 6, 2]
quantum = 5

waiting_time, turnaround_time, avg_waiting_time, avg_turnaround_time = round_robin(processes, burst_time, quantum)

print("Processes:", processes)
print("Burst Time:", burst_time)
print("Waiting Time:", waiting_time)
print("Turnaround Time:", turnaround_time)
print("Average Waiting Time:", avg_waiting_time)
print("Average Turnaround Time:", avg_turnaround_time)
```
